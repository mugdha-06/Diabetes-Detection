Importing the dependencies

```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn import svm
from sklearn.metrics import accuracy_score
from sklearn.preprocessing import LabelEncoder
```

Data Collection
(Diabetes prediction dataset from kaggle)


```python
diabetes_data = pd.read_csv('/content/diabetes_prediction_dataset.csv')
```

```python
diabetes_data.head()
```

```python
#rows and cols of the dataset
diabetes_data.shape
```

```python
#statistical info of the data
diabetes_data.describe()
```

```python
diabetes_data.isnull().sum()
```

```python
diabetes_data['diabetes'].value_counts()
```

```python
#0 -> non diabetic 1->Diabetic
diabetes_data['smoking_history'].value_counts()
```

```python
diabetes_data = pd.get_dummies(
    diabetes_data,
    columns=['smoking_history'],
    drop_first=True,
    dtype=int
)
```

```python
diabetes_data.head()
```

Label Encoding


```python
le_gender = LabelEncoder()
diabetes_data['gender'] = le_gender.fit_transform(diabetes_data['gender'])
```

```python
diabetes_data.head() #0->female 1->male
print(diabetes_data.shape)
```

Separating Data and labels


```python
X= diabetes_data.drop(columns = 'diabetes', axis =1)
Y= diabetes_data['diabetes']
```

Data Standardization

```python
scaler = StandardScaler()
scaler.fit(X)
```

```python
scalerized_data = scaler.transform(X)
```

```python
print(scalerized_data)
```

```python
X= scalerized_data
Y= diabetes_data['diabetes']
```

Test Train Split

```python
X_train,X_test,Y_train,Y_test = train_test_split(X,Y,test_size=0.2,stratify=Y,random_state=2)
```

```python
print(X.shape,X_train.shape,X_test.shape)
```

Model Training

```python
classifier = svm.SVC(kernel='linear')
```

```python
#training the svm classifier
classifier.fit(X_train,Y_train)
```

Model Evaluation

```python
#Accuracy Score
X_train_prediction = classifier.predict(X_train)
training_data_accuracy = accuracy_score(X_train_prediction,Y_train)
```

```python
print('Accuracy score of the training data :',training_data_accuracy)
```

```python
X_test_prediction = classifier.predict(X_test)
testing_data_accuracy = accuracy_score(X_test_prediction,Y_test)
```

```python
print('Accuracy score of the testing data :',testing_data_accuracy)
```

Making a predictive system

```python
new_patient = {
    'gender': 'Male',             # 'Female' or 'Male'
    'age': 57.0,
    'hypertension': 1,              # 0 = No, 1 = Yes
    'heart_disease': 0,             # 0 = No, 1 = Yes
    'bmi': 23.10,
    'HbA1c_level': 5.5,
    'blood_glucose_level': 113,
    'smoking_history': 'never'      # Matches the exact raw categories
}

input_df = pd.DataFrame([new_patient])

# Map gender manually to match 0 -> Female, 1 -> Male consistently
gender_map = {'Female': 0, 'Male': 1}
input_df['gender'] = input_df['gender'].map(gender_map)

# One-hot encode categorical smoking history columns
input_df = pd.get_dummies(
    input_df,
    columns=['smoking_history'],
    dtype=int
)

# Re-align input_df columns to match the features used to train the scaler and classifier
feature_columns = diabetes_data.drop(columns=['diabetes']).columns
input_df = input_df.reindex(columns=feature_columns, fill_value=0)

# Standardize and predict
input_data = scaler.transform(input_df)
prediction = classifier.predict(input_data)
```

```python
if prediction[0] == 0:
    print('The person is not diabetic')
else:
    print('The person is diabetic')
```
